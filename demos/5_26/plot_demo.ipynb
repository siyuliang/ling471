```python
## Set up code
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
## Creating a line plot
x = np.linspace(0, 2*np.pi, num=200)
y = np.sin(x)

df = pd.DataFrame({'x': x, 'y': y})
df.plot('x', 'y')
plt.show()
```

```python
x = np.linspace(0, 10, num=200)
y = x**2

df = pd.DataFrame({'x': x, 'y': y})
df.plot('x', 'y')
plt.show()
```

```python
## Making a scatterplot
x = np.random.rand(50)
y = np.random.rand(50)

df = pd.DataFrame({'x': x, 'y': y})
df.plot.scatter('x', 'y')
plt.show()
```

```python
## Ordering for scatter vs. line
x = np.random.rand(200)*2*np.pi
y = np.sin(x)

df = pd.DataFrame({'x': x, 'y': y})
df.plot('x', 'y')
plt.show()
```

```python
x = np.random.rand(200)*2*np.pi
y = np.sin(x)

df = pd.DataFrame({'x': x, 'y': y})
df.plot.scatter('x', 'y')
plt.show()
```

```python
## Fixing the line plot
## For a line plot, sort by the x-axis variable first.x = np.random.rand(200)*2*np.pi
y = np.sin(x)

df = pd.DataFrame({'x': x, 'y': y})
df = df.sort_values('x')

df.plot('x', 'y')
plt.show()
```

```python
## Making a barplot
df = pd.DataFrame({'word': ['the', 'cute', 'cat'], 'count': [9876, 654, 320]})

n_col = len(set(df.word))
colors = [plt.cm.Set2(x) for x in range(n_col)]

df.plot.bar('word', 'count', color=colors)
plt.legend('', frameon=False)
plt.subplots_adjust(bottom=0.15)
plt.show()
```

```python
## Making a heatmap
a = np.zeros((100, 100))

for i in range(a.shape[0]):
    for j in range(a.shape[1]):
        a[i, j] = np.linalg.norm([i-49, j-49])

plt.pcolormesh(a)
plt.colorbar()
plt.show()
```
